Assemble dashboard datasets for Power BI/Tableau.


```python
import pandas as pd
import matplotlib.pyplot as plt

# Example data for a quick bar chart (replace with your processed dataset as needed)
data = pd.DataFrame({
	"Category": ["iPhone", "iPad", "Mac", "Watch", "AirPods"],
	"Revenue": [120, 60, 85, 40, 55]
})

ax = data.set_index("Category").plot(kind="bar", legend=False, figsize=(8, 4), color="#4F46E5")
ax.set_ylabel("Revenue (units)")
ax.set_title("Sample Dashboard Bar Chart")
plt.tight_layout()
plt.show()
```
